#Comparing performance between CPU and GPU

```python
import torch
import time
```

## Change the runtime type to use T4 GPU

```python
#Check if the device is "cuda" if not "cpu"
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using {device}")
```

## Time Performance of CPU

```python
size = 5000

a_cpu = torch.randn(size, size)
b_cpu = torch.randn(size, size)

#print(a_cpu)
#print(b_cpu)

start = time.time()

c_cpu = a_cpu @ b_cpu
#print(c_cpu)

cpu_time = time.time() - start

print(f"CPU time: {cpu_time:.3f}s")
```

## Time Performance of GPU

```python
if torch.cuda.is_available():
    a_gpu = a_cpu.to('cuda')
    b_gpu = b_cpu.to('cuda')

    # Without synchronization, it will load up the rest of the calculation time
    torch.cuda.synchronize()

    start = time.time()

    c_gpu = a_gpu @ b_gpu

    torch.cuda.synchronize()

    gpu_time = time.time() - start

    print(f'GPU time: {gpu_time:.3f}s')

    print(f'GPU speedup: {cpu_time / gpu_time:.0f}x')

```
